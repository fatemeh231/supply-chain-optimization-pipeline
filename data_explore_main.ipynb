```python
# 01_data_exploration.ipynb
# Supply Chain Logistics Problem - Data Exploration

import pandas as pd
import os

# Path to the Excel file
file_path = "Supply chain logisitcs problem.xlsx"

# Check if file exists
if os.path.exists(file_path):
    print("✅ File found!")
else:
    print("❌ File not found. Check the path.")
    print(f"Looking for: {file_path}")

# Load all 7 sheets
print("\n" + "="*60)
print("📊 LOADING ALL SHEETS")
print("="*60)

order_list = pd.read_excel(file_path, sheet_name="OrderList")
freight_rates = pd.read_excel(file_path, sheet_name="FreightRates")
wh_costs = pd.read_excel(file_path, sheet_name="WhCosts")
wh_capacities = pd.read_excel(file_path, sheet_name="WhCapacities")
products_per_plant = pd.read_excel(file_path, sheet_name="ProductsPerPlant")
vmi_customers = pd.read_excel(file_path, sheet_name="VmiCustomers")
plant_ports = pd.read_excel(file_path, sheet_name="PlantPorts")

print("✅ All 7 sheets loaded successfully!")
```

```python
# Dictionary of all sheets
sheets = {
    "OrderList": order_list,
    "FreightRates": freight_rates,
    "WhCosts": wh_costs,
    "WhCapacities": wh_capacities,
    "ProductsPerPlant": products_per_plant,
    "VmiCustomers": vmi_customers,
    "PlantPorts": plant_ports
}

for name, df in sheets.items():
    print("\n" + "="*60)
    print(f"📄 Sheet: {name}")
    print("="*60)
    print(f"   Shape: {df.shape[0]} rows, {df.shape[1]} columns")
    print(f"   Columns: {df.columns.tolist()}")
    print(f"\n   First 5 Rows:")
    print(df.head())
    print(f"\n   Null Values:")
    print(df.isnull().sum())
```

```python
# Investigate V44_3
print("\n" + "="*60)
print("🔍 INVESTIGATING V44_3 CARRIER")
print("="*60)

# Check if V44_3 exists in OrderList
v44_3_orders = order_list[order_list['Carrier'] == 'V44_3']
print(f"Orders with V44_3: {len(v44_3_orders)}")

# Check if V44_3 exists in FreightRates
v44_3_rates = freight_rates[freight_rates['Carrier'] == 'V44_3']
print(f"V44_3 in FreightRates: {len(v44_3_rates)}")

if len(v44_3_rates) == 0:
    print("⚠️ V44_3 is NOT in FreightRates!")
    print("   This means it's a DISCONTINUED CARRIER.")
    print("   You need to decide how to handle this in your analysis.")
else:
    print("✅ V44_3 found in FreightRates.")

# Show sample of V44_3 orders
if len(v44_3_orders) > 0:
    print(f"\n📋 Sample V44_3 Orders:")
    print(v44_3_orders[['Order ID', 'Carrier', 'Origin Port', 'Destination Port', 'Weight']].head())
```

```python
print("\n" + "="*60)
print("🔗 TABLE RELATIONSHIPS")
print("="*60)

# 1. OrderList ↔ FreightRates
print("\n📦 OrderList ↔ FreightRates:")
print(f"   Carriers in OrderList: {order_list['Carrier'].nunique()}")
print(f"   Carriers in FreightRates: {freight_rates['Carrier'].nunique()}")

# 2. Check Plants
print("\n🏭 Plants:")
print(f"   Plants in OrderList: {order_list['Plant Code'].nunique()}")
print(f"   Plants in PlantPorts: {plant_ports['Plant Code'].nunique()}")

# 3. Check Products
print("\n📦 Products:")
print(f"   Products in OrderList: {order_list['Product ID'].nunique()}")
print(f"   Products in ProductsPerPlant: {products_per_plant['Product ID'].nunique()}")
```

```python
print("\n" + "="*60)
print("🔍 CHECKING DATA COMPLETENESS")
print("="*60)

# 1. Check Plants
plants_in_orders = set(order_list['Plant Code'].unique())
plants_in_wc = set(wh_costs['WH'].unique())
plants_in_cap = set(wh_capacities['Plant ID'].unique())
plants_in_pp = set(products_per_plant['Plant Code'].unique())

print("\n🏭 Plants:")
print(f"   In OrderList: {len(plants_in_orders)}")
print(f"   In WhCosts: {len(plants_in_wc)}")
print(f"   In WhCapacities: {len(plants_in_cap)}")
print(f"   In ProductsPerPlant: {len(plants_in_pp)}")

# Check which plants are missing
missing_wc = plants_in_orders - plants_in_wc
missing_cap = plants_in_orders - plants_in_cap
missing_pp = plants_in_orders - plants_in_pp

if missing_wc:
    print(f"\n   ⚠️ Plants missing in WhCosts: {missing_wc}")
if missing_cap:
    print(f"   ⚠️ Plants missing in WhCapacities: {missing_cap}")
if missing_pp:
    print(f"   ⚠️ Plants missing in ProductsPerPlant: {missing_pp}")

# 2. Check Products
products_in_orders = set(order_list['Product ID'].unique())
products_in_pp = set(products_per_plant['Product ID'].unique())

print("\n📦 Products:")
print(f"   In OrderList: {len(products_in_orders)}")
print(f"   In ProductsPerPlant: {len(products_in_pp)}")

missing_products = products_in_orders - products_in_pp
if missing_products:
    print(f"   ⚠️ Products missing in ProductsPerPlant: {len(missing_products)}")
```

```python
print("\n" + "="*60)
print("🔧 BUILDING MASTER DATASET")
print("="*60)

# Step 1: Flag V44_3 in OrderList
order_list['carrier_status'] = order_list['Carrier'].apply(
    lambda x: 'Discontinued' if x == 'V44_3' else 'Active'
)

print(f"\n✅ Carrier status added:")
print(order_list['carrier_status'].value_counts())

# Step 2: Merge OrderList with FreightRates
# We need to match on Carrier, Origin Port, Destination Port, and Weight range
print("\n🔄 Merging OrderList with FreightRates...")

# First, create a copy of OrderList
master_df = order_list.copy()

# For each order, find the matching freight rate
# We'll do this step by step

# Step 2a: Merge on Carrier, Origin Port, Destination Port
master_df = master_df.merge(
    freight_rates,
    left_on=['Carrier', 'Origin Port', 'Destination Port'],
    right_on=['Carrier', 'orig_port_cd', 'dest_port_cd'],
    how='left'
)

print(f"   After merging: {len(master_df)} rows, {len(master_df.columns)} columns")

# Step 2b: Check which orders got rates
matched = master_df['rate'].notna().sum()
total = len(master_df)
print(f"   Orders with rates: {matched} ({matched/total*100:.2f}%)")
print(f"   Orders without rates: {total - matched} ({(total-matched)/total*100:.2f}%)")

# Step 3: Check V44_3 specifically
v44_3_matched = master_df[(master_df['Carrier'] == 'V44_3') & (master_df['rate'].notna())]
print(f"\n   V44_3 orders with rates: {len(v44_3_matched)} (should be 0!)")

# Step 4: Add plant information
print("\n🔄 Merging with Plant Ports...")
master_df = master_df.merge(
    plant_ports,
    left_on=['Plant Code'],
    right_on=['Plant Code'],
    how='left'
)
print(f"   After merging PlantPorts: {len(master_df)} rows")

# Step 5: Add warehouse costs
print("\n🔄 Merging with Warehouse Costs...")
master_df = master_df.merge(
    wh_costs,
    left_on=['Plant Code'],
    right_on=['WH'],
    how='left'
)
print(f"   After merging WhCosts: {len(master_df)} rows")

# Step 6: Add warehouse capacities
print("\n🔄 Merging with Warehouse Capacities...")
master_df = master_df.merge(
    wh_capacities,
    left_on=['Plant Code'],
    right_on=['Plant ID'],
    how='left'
)
print(f"   After merging WhCapacities: {len(master_df)} rows")

# Step 7: Add product-plant mapping
print("\n🔄 Merging with Products Per Plant...")
master_df = master_df.merge(
    products_per_plant,
    left_on=['Plant Code', 'Product ID'],
    right_on=['Plant Code', 'Product ID'],
    how='left'
)
print(f"   After merging ProductsPerPlant: {len(master_df)} rows")

# Step 8: Add VMI customer information
print("\n🔄 Merging with VMI Customers...")
master_df = master_df.merge(
    vmi_customers,
    left_on=['Plant Code'],
    right_on=['Plant Code'],
    how='left'
)
print(f"   After merging VmiCustomers: {len(master_df)} rows")

# Step 9: Calculate freight cost
print("\n🔄 Calculating Freight Cost...")
master_df['freight_cost'] = master_df.apply(
    lambda row: row['Weight'] * row['rate'] if pd.notna(row['rate']) else None,
    axis=1
)

# Step 10: Calculate total cost (freight + warehouse)
master_df['total_cost'] = master_df.apply(
    lambda row: row['freight_cost'] + (row['Unit quantity'] * row['Cost/unit']) 
    if pd.notna(row['freight_cost']) and pd.notna(row['Cost/unit']) 
    else None,
    axis=1
)

print("\n" + "="*60)
print("📊 MASTER DATASET SUMMARY")
print("="*60)
print(f"   Total Rows: {len(master_df)}")
print(f"   Total Columns: {len(master_df.columns)}")
print(f"   Orders with freight cost: {master_df['freight_cost'].notna().sum()}")
print(f"   Orders without freight cost: {master_df['freight_cost'].isna().sum()}")
print(f"   V44_3 orders: {len(master_df[master_df['Carrier'] == 'V44_3'])}")

# Show sample of V44_3 rows
print("\n📋 V44_3 Sample (should show None/NaN for rate and freight_cost):")
v44_3_sample = master_df[master_df['Carrier'] == 'V44_3'][['Order ID', 'Carrier', 'Weight', 'rate', 'freight_cost', 'carrier_status']].head(5)
print(v44_3_sample)
```

```python
# Save master dataset as Parquet
print("\n" + "="*60)
print("💾 SAVING MASTER DATASET")
print("="*60)

# Define output path
output_path = "supply_chain_master.parquet"

# Save as Parquet
master_df.to_parquet(output_path, index=False)
print(f"✅ Master dataset saved to: {output_path}")
print(f"   Shape: {master_df.shape[0]} rows, {master_df.shape[1]} columns")
print(f"   Size: {os.path.getsize(output_path) / 1024 / 1024:.2f} MB")
```
